```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('usa_traffic_accidents_2023_2024_1000.csv')

# 1. Accidents by Day of Week
plt.figure(figsize=(8,5))
df['Day_of_Week'].value_counts().plot(
    kind='bar',
    color=['red','blue','green','orange','purple','cyan','magenta']
)
plt.title("Accidents by Day of Week")
plt.xlabel("Day")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# 2. Accident Severity Distribution
plt.figure(figsize=(6,5))
df['Accident_Severity'].value_counts().plot(
    kind='bar',
    color=['#1f77b4','#ff7f0e','#2ca02c']   # custom hex colors
)
plt.title("Accident Severity Distribution")
plt.xlabel("Severity")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# 3. Weather condition vs accidents
plt.figure(figsize=(10,5))
df['Weather_Condition'].value_counts().plot(
    kind='bar',
    color=['#e6194B','#3cb44b','#ffe119','#4363d8','#f58231','#911eb4']
)
plt.title("Accidents by Weather Condition")
plt.xlabel("Weather")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# 4. Accidents by Light Condition
plt.figure(figsize=(8,5))
df['Light_Condition'].value_counts().plot(
    kind='bar',
    color=['#42d4f4','#f032e6','#fabebe','#469990']
)
plt.title("Accidents by Light Condition")
plt.xlabel("Light Condition")
plt.ylabel("Count")
plt.tight_layout()
plt.show()







```
